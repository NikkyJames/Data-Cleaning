### Missing Data Handling Strategy

After standardising the dataset, both actual missing values and invalid placeholder values (ERROR and UNKNOWN) were treated as missing values.

The following strategies were selected based on the data type and characteristics of each column:

Item: Missing categorical values will be replaced using the mode, as Item is a categorical variable and the most frequently occurring valid item provides a reasonable representative value.
Quantity: Missing values will be replaced using the median. The valid quantities range from 1 to 5, with a median of 3. Median imputation is appropriate because it is less sensitive to potential outliers.
Price Per Unit: Missing values will be replaced using the median of 3. This preserves a typical price without being unduly influenced by extreme values.
Total Spent: Missing values will be handled using the relationship between Quantity and Price Per Unit where possible. Since Total Spent represents transaction spending, values will be derived from available transaction information rather than relying solely on a simple statistical imputation.
Payment Method: Missing categorical values will be replaced using the mode, avoiding the loss of a large number of transaction records.
Location: Missing categorical values will be replaced using the mode, rather than deleting a substantial proportion of the dataset.
Transaction Date: Missing dates will be handled separately after conversion to datetime, using an appropriate date-based strategy rather than applying numeric or categorical imputation.
This approach preserves as many records as possible while maintaining appropriate data types and reducing the risk of introducing unrealistic values.

```python
# Calculate imputation values from the cleaned working dataset

# Numeric columns: median
numeric_medians = df_clean[["Quantity", "Price Per Unit", "Total Spent"]].median()

# Categorical columns: mode
categorical_modes = df_clean[["Item", "Payment Method", "Location"]].mode().iloc[0]

print("=== MEDIAN VALUES ===")
print(numeric_medians)

print("\n=== MODE VALUES ===")
print(categorical_modes)
```

```python
# Calculate statistics for missing-data decisions

# Numeric columns: mean and median
numeric_summary = df_clean[["Quantity", "Price Per Unit", "Total Spent"]].agg(
    ["mean", "median"]
)

# Categorical columns: mode
categorical_modes = df_clean[["Item", "Payment Method", "Location"]].mode().iloc[0]

print("=== MEAN AND MEDIAN VALUES ===")
print(numeric_summary)

print("\n=== MODE VALUES ===")
print(categorical_modes)
```

```python
# Impute missing values in categorical and numeric columns

# Categorical columns: mode
df_clean["Item"] = df_clean["Item"].fillna("Juice")
df_clean["Payment Method"] = df_clean["Payment Method"].fillna("Digital Wallet")
df_clean["Location"] = df_clean["Location"].fillna("Takeaway")

# Numeric columns: median
df_clean["Quantity"] = df_clean["Quantity"].fillna(3)
df_clean["Price Per Unit"] = df_clean["Price Per Unit"].fillna(3)

print("Missing values after categorical and numeric imputation:")
print(df_clean.isnull().sum())
```

```python
# Calculate Total Spent from Quantity × Price Per Unit

missing_total = df_clean["Total Spent"].isna()

df_clean.loc[missing_total, "Total Spent"] = (
    df_clean.loc[missing_total, "Quantity"] *
    df_clean.loc[missing_total, "Price Per Unit"]
)

print("Missing Total Spent values after calculation:")
print(df_clean["Total Spent"].isnull().sum())
```

```python
# Convert Transaction Date to datetime

df_clean["Transaction Date"] = pd.to_datetime(
    df_clean["Transaction Date"],
    errors="coerce"
)

print("Transaction Date data type:")
print(df_clean["Transaction Date"].dtype)

print("\nMissing Transaction Date values:")
print(df_clean["Transaction Date"].isnull().sum())
```

```python
# Find the most frequent valid transaction date

date_mode = df_clean["Transaction Date"].mode()[0]

print("Most frequent transaction date:", date_mode)
print("Frequency:", df_clean["Transaction Date"].value_counts().iloc[0])
```

### Transaction Date Imputation Decision

The Transaction Date column contained 460 missing or invalid values after ERROR and UNKNOWN entries were standardised as missing values.

Because Transaction Date is a datetime variable, numeric mean or median imputation was not appropriate. Row deletion was also avoided because it would unnecessarily remove 460 valid transaction records.

The mode date, 2023-02-06, was selected for imputation because it was the most frequently occurring valid transaction date, with 40 observations. This approach preserves all transaction records while assigning a valid datetime value to the missing entries.

A limitation of this method is that it may create an artificial concentration of transactions on the mode date. However, it is appropriate for this cleaning exercise because the assignment permits mode imputation and the original dates cannot be reliably reconstructed from the available information.

```python
# Impute missing Transaction Date values using the mode

df_clean["Transaction Date"] = df_clean["Transaction Date"].fillna(date_mode)

print("Missing Transaction Date values after imputation:")
print(df_clean["Transaction Date"].isnull().sum())
```

```python
# Identify and remove duplicate rows

duplicates_before = df_clean.duplicated().sum()

df_clean = df_clean.drop_duplicates().copy()

duplicates_after = df_clean.duplicated().sum()

print("Duplicate rows before removal:", duplicates_before)
print("Duplicate rows after removal:", duplicates_after)
print("Rows after duplicate removal:", len(df_clean))
```

### Standardisation — Observation

```python
# Detect outliers using the IQR method

numeric_columns = ["Quantity", "Price Per Unit", "Total Spent"]

for col in numeric_columns:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df_clean[
        (df_clean[col] < lower_bound) | 
        (df_clean[col] > upper_bound)
    ]
    
    print(f"\n{col}")
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of outliers:", len(outliers))
```

```python
# Inspect Total Spent outliers

Q1 = df_clean["Total Spent"].quantile(0.25)
Q3 = df_clean["Total Spent"].quantile(0.75)
IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR

total_spent_outliers = df_clean[
    df_clean["Total Spent"] > upper_bound
]["Total Spent"]

print("Number of Total Spent outliers:", len(total_spent_outliers))
print("\nOutlier value counts:")
print(total_spent_outliers.value_counts().sort_index())

print("\nMinimum outlier value:", total_spent_outliers.min())
print("Maximum outlier value:", total_spent_outliers.max())
```

### Outlier Detection — Observation

The IQR method was used to identify potential outliers in the numeric columns.

Quantity and Price Per Unit had no detected outliers. Total Spent had 268 observations identified as potential outliers because the IQR method produced an upper bound of 24.0.

Further inspection showed that all 268 flagged observations had a Total Spent value of 25.0. Since 25.0 is a valid transaction amount and there is no indication that these values are data-entry errors, the observations were retained rather than removed or capped.

This decision preserves legitimate high-value transactions and avoids distorting the original spending information.

```python
# Count transactions where Total Spent is inconsistent
# with Quantity × Price Per Unit

inconsistent_count = total_mismatch.sum()

print("Total inconsistent Total Spent values:", inconsistent_count)
print("Percentage of dataset affected:",
      round((inconsistent_count / len(df_clean)) * 100, 2), "%")
```

### Total Spent Consistency Check — Observation

A consistency check was performed by comparing `Total Spent` with the expected transaction amount calculated as `Quantity × Price Per Unit`.

The check identified **733 inconsistent transactions**, representing **7.33% of the dataset**. The mismatches occurred across **54 different value patterns**, indicating that the issue is not limited to a single unusual value.

Because `Total Spent` should represent the amount spent on each transaction, the expected value can be reliably derived from the available `Quantity` and `Price Per Unit` fields. Therefore, the inconsistent `Total Spent` values will be corrected using:

**Total Spent = Quantity × Price Per Unit**

This correction improves internal consistency and ensures that the monetary value accurately reflects the transaction quantity and unit price.

```python
# Correct inconsistent Total Spent values
# using Quantity × Price Per Unit

df_clean.loc[total_mismatch, "Total Spent"] = (
    df_clean.loc[total_mismatch, "Quantity"] *
    df_clean.loc[total_mismatch, "Price Per Unit"]
)

print("Total Spent values corrected:", total_mismatch.sum())
```

```python
# Verify Total Spent consistency after correction

calculated_total = (
    df_clean["Quantity"] * df_clean["Price Per Unit"]
)

total_mismatch_after = ~np.isclose(
    df_clean["Total Spent"],
    calculated_total
)

print("Total Spent mismatches after correction:", total_mismatch_after.sum())
```

```python
# Re-check outliers in Total Spent after consistency correction

Q1 = df_clean["Total Spent"].quantile(0.25)
Q3 = df_clean["Total Spent"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

total_spent_outliers = df_clean[
    (df_clean["Total Spent"] < lower_bound) |
    (df_clean["Total Spent"] > upper_bound)
]

print("=== TOTAL SPENT OUTLIER CHECK AFTER CORRECTION ===")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of outliers:", len(total_spent_outliers))

print("\nOutlier value counts:")
print(total_spent_outliers["Total Spent"].value_counts().sort_index())
```
